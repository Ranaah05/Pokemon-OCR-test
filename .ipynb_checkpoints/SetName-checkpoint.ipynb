```python
import os
import re
import requests
from typing import Dict, Optional

class FinalCardPricer:
    def __init__(self):
        # API key for PriceTracker, count API calls for tracking
        self.price_api_key = os.getenv("POKE_PRICE_TRACKER_API_KEY")
        self.api_calls_made = 0
        
    #************************************************************
    # Method: parse_card_input
    # Description:
    #   Parse user input to extract card name and number
    #   Examples:
    #   "Rayquaza V 194/203" → ("Rayquaza V", "194")
    #   "Cynthia SV82" → ("Cynthia", "SV82")
    #************************************************************
    def parse_card_input(self, card_input: str) -> tuple:
        parts = card_input.strip().split()
        if not parts:
            return card_input, None
            
        last_part = parts[-1]
        
        if '/' in last_part and re.search(r'\d+', last_part):  # Format "194/203"
            number = last_part.split('/')[0]
            name = ' '.join(parts[:-1])
            return name, number
        
        if re.match(r'^[A-Z]*\d+$', last_part):  # Format "SV82", "25"
            number = last_part
            name = ' '.join(parts[:-1]) if len(parts) > 1 else card_input
            return name, number
        
        return card_input, None  # Only name, no number
    
    #************************************************************
    # Method: get_card_from_pokemontcg
    # Description:
    #   Query pokemontcg.io API for exact card details
    #   Returns first matching card
    #************************************************************
    def get_card_from_pokemontcg(self, name: str, number: str = None) -> Optional[Dict]:
        print(f"\n--- Step 1: pokemontcg.io search ---")
        print(f"Searching for: {name} #{number}")
        
        query_parts = [f'name:\"{name}\"']
        if number:
            query_parts.append(f'number:{number}')
        query = ' AND '.join(query_parts)
        
        try:
            response = requests.get(
                "https://api.pokemontcg.io/v2/cards",
                params={"q": query, "pageSize": 5},
                timeout=30
            )
            response.raise_for_status()
            cards = response.json().get("data", [])
            
            if not cards:
                print("   No cards found")
                return None
            
            print(f"   Found {len(cards)} matches:")
            for i, card in enumerate(cards):          # Loop through results
                card_name = card.get('name', '')      # Card name
                card_number = card.get('number', '')  # Card number
                set_info = card.get('set', {})        # Set info dict
                set_name = set_info.get('name', 'Unknown')  # Set name
                print(f"     {i+1}. {card_name} #{card_number} from {set_name}")
            
            best_card = cards[0]  # Pick first result
            print(f"   Selected: {best_card['name']} #{best_card['number']}")
            return best_card
            
        except Exception as e:
            print(f"   Error: {e}")
            return None
    
    #************************************************************
    # Method: matches_target_card
    # Description:
    #   Compare a PriceTracker result to the target card
    #   Returns (is_match, quality_score)
    #************************************************************
    def matches_target_card(self, target_name: str, target_number: str, 
                           result_name: str, result_info: dict) -> tuple:
        def get_base_name(name):  # Remove extra text in parentheses
            return re.sub(r'\s*\([^)]+\)', '', name).strip().lower()
        
        target_base = get_base_name(target_name)
        result_base = get_base_name(result_name)
        
        match_quality = 0
        if target_base != result_base:  # Names must match
            return False, 0
        match_quality += 10
        
        if target_number:  # If number is provided, check match
            result_number = str(result_info.get('number', ''))
            if target_number in result_number:
                match_quality += 20
            elif target_number.startswith('SV') and 'SV' in result_number:
                match_quality += 15
        
        if target_name.lower() == result_name.lower():  # Exact match bonus
            match_quality += 5
        
        return True, match_quality
    
    #************************************************************
    # Method: search_price_tracker
    # Description:
    #   Query PriceTracker API and select best card match
    #************************************************************
    def search_price_tracker(self, target_name: str, target_number: str = None) -> Optional[Dict]:
        print(f"\n--- Step 2: PriceTracker search ---")
        print(f"Looking for: {target_name} #{target_number}")
        
        headers = {"Authorization": f"Bearer {self.price_api_key}"}
        search_terms = [target_name]
        if target_number:
            search_terms.insert(0, f"{target_name} {target_number}")
        
        for i, search_term in enumerate(search_terms):
            print(f"   Attempt {i+1}: {search_term}")
            params = {"search": search_term, "limit": 10, "includeHistory": "false"}
            
            try:
                response = requests.get(
                    "https://www.pokemonpricetracker.com/api/v2/cards",
                    headers=headers,
                    params=params,
                    timeout=30
                )
                self.api_calls_made += 1
                remaining = max(0, 200 - self.api_calls_made)
                print(f"   API Call #{self.api_calls_made} | ~{remaining} remaining")
                
                if response.status_code == 200:
                    results = response.json().get("data", [])
                    if not results:
                        print("   No results found")
                        continue
                    
                    print(f"   Found {len(results)} results:")
                    for j, result in enumerate(results):   # Show result list
                        name = result.get("name", "")
                        number = result.get("number", "")
                        set_info = result.get("set", {})
                        set_name = set_info.get("name", "Unknown") if isinstance(set_info, dict) else str(set_info)
                        price = result.get("prices", {}).get("market", "N/A")
                        print(f"     {j+1}. {name} #{number} | {set_name} | ${price}")
                    
                    # Pick best match
                    best_match = None
                    best_quality = 0
                    print(f"\n   Checking matches...")
                    
                    for result in results:
                        result_name = result.get("name", "")
                        is_match, quality = self.matches_target_card(
                            target_name, target_number, result_name, result
                        )
                        print(f"     {result_name} | Match: {is_match} | Quality: {quality}")
                        if is_match and quality > best_quality:
                            best_match = result
                            best_quality = quality
                            print("       New best match")
                    
                    if best_match:
                        price = best_match.get("prices", {}).get("market")
                        print(f"\n   BEST MATCH: {best_match.get('name')} | ${price}")
                        return {
                            "card": best_match,
                            "price": price,
                            "search_term": search_term,
                            "match_quality": best_quality
                        }
                    else:
                        print("   No suitable matches found")
                        
                elif response.status_code == 429:
                    print("   Rate limit exceeded")
                    return None
                else:
                    print(f"   API Error: {response.status_code}")
                    
            except Exception as e:
                print(f"   Request failed: {e}")
        
        return None
    
    #************************************************************
    # Method: get_card_price
    # Description:
    #   Full lookup process:
    #   - Parse input
    #   - Search pokemontcg.io
    #   - Search PriceTracker
    #   - Return final result
    #************************************************************
    def get_card_price(self, card_input: str) -> Optional[Dict]:
        print("="*60)
        print(f"SEARCH: {card_input}")
        print("="*60)
        
        name, number = self.parse_card_input(card_input)  # Parse input
        print(f"Parsed -> Name: '{name}', Number: '{number}'")
        
        card_data = self.get_card_from_pokemontcg(name, number)  # pokemontcg.io
        if not card_data:
            print("Card not found in pokemontcg.io")
            return None
        
        exact_name = card_data.get('name')
        exact_number = card_data.get('number')
        
        price_result = self.search_price_tracker(exact_name, exact_number)  # PriceTracker
        if price_result:
            print("\n" + "="*60)
            print("SUCCESS")
            print(f"Input: {card_input}")
            print(f"Card: {exact_name} #{exact_number}")
            print(f"Price: ${price_result['price']}")
            print(f"Quality: {price_result['match_quality']}")
            print(f"API Calls: {self.api_calls_made}")
            print("="*60)
            return price_result
        else:
            print("No price found")
            return None

#************************************************************
# Function: get_card_price (simple interface)
# Description:
#   Returns just the price for a given card input
#************************************************************
def get_card_price(card_input: str) -> Optional[float]:
    pricer = FinalCardPricer()
    result = pricer.get_card_price(card_input)
    if result and result.get('price'):
        return float(result['price'])
    return None

#************************************************************
# MAIN (testing)
#************************************************************
if __name__ == "__main__":
    pricer = FinalCardPricer()
    test_cards = ["Rayquaza V 194/203", "Cynthia SV82"]
    for card in test_cards:
        result = pricer.get_card_price(card)
        print(f"\nFinal result for {card}: ${result['price'] if result else 'Not found'}")
        print("-" * 50)
```
