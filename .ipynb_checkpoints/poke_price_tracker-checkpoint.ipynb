```python

import os
import requests

#  Load API key from environment variable
API_KEY = os.getenv("POKE_PRICE_TRACKER_API_KEY")

# Check if API key is loaded
if not API_KEY:
    print("Error: API key not found. Please set POKE_PRICE_TRACKER_API_KEY.")
```

```python
# Step 2: Function to fetch card price
def fetch_card_price(card_id):
    url = f"https://www.pokemonpricetracker.com/api/v2/cards/{card_id}"
    headers = {
        "Authorization": f"Bearer {API_KEY}"
    }
    
    response = requests.get(url, headers=headers)
    if response.status_code == 200:
        return response.json()
    else:
        print("Error:", response.status_code, response.text)
        return None

# Step 3: Test with Rayquaza C from Supreme Victors
card_id = "pl3-8"   # From your TCG.io output
card_data = fetch_card_price(card_id)

# Step 4: Display results in a simple way
if card_data:
    print("Card Name:", card_data["data"]["name"])
    print("Set Name:", card_data["data"]["set"]["name"])
    
    pricing = card_data["data"].get("pricing", {})
    print("\nPricing:")
    for market, info in pricing.items():
        print(f"- {market}: {info}")
```
