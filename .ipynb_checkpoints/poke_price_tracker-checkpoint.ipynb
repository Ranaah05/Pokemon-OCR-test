```python

import os
import requests

#  Load API key from environment variable
API_KEY = os.getenv("POKE_PRICE_TRACKER_API_KEY")

# Check if API key is loaded
if not API_KEY:
    print("Error: API key not found. Please set POKE_PRICE_TRACKER_API_KEY.")
```

```python
import os
import re
import time
import requests
from typing import Dict, Optional, Tuple

IOAPI_KEY = "ac4aadab-8d33-45ff-8ca2-666723786cfc"  

class FinalCardPricer:
    def __init__(self):
        # You'll need separate API keys for different services
        self.pokemontcg_api_key = IOAPI_KEY  # For pokemontcg.io
        self.pricetracker_api_key = API_KEY  # For pokemonpricetracker.com
        self.api_calls_made = 0

    def parse_card_input(self, card_input: str) -> Tuple[str, Optional[str]]:
        """
        Parse card input to extract name and number.
        Examples:
        "Rayquaza V 194/203" -> ("Rayquaza V", "194")
        "Cynthia SV82" -> ("Cynthia", "SV82")
        """
        parts = card_input.strip().split()
        if not parts:
            return card_input, None

        last_part = parts[-1]

        if '/' in last_part and re.search(r'\d+', last_part):
            number = last_part.split('/')[0]
            name = ' '.join(parts[:-1])
            return name, number

        if re.match(r'^[A-Z]*\d+$', last_part):
            number = last_part
            name = ' '.join(parts[:-1]) if len(parts) > 1 else card_input
            return name, number

        return card_input, None

    def get_card_from_pokemontcg(self, name: str, number: Optional[str] = None, retries: int = 3) -> Optional[Dict]:
        """
        Get card details from pokemontcg.io, retrying on 504 Gateway Timeout.
        """
        query_parts = [f'name:"{name}"']
        if number:
            query_parts.append(f'number:{number}')
        query = ' AND '.join(query_parts)

        url = "https://api.pokemontcg.io/v2/cards"
        headers = {"X-Api-Key": self.pokemontcg_api_key}

        for attempt in range(retries):
            try:
                response = requests.get(url, headers=headers, params={"q": query, "pageSize": 5}, timeout=30)
                if response.status_code == 504:
                    # Retry on 504
                    print(f"504 Gateway Timeout, retrying ({attempt+1}/{retries})...")
                    time.sleep(2)
                    continue
                response.raise_for_status()
                data = response.json().get("data", [])
                if not data:
                    return None
                # Return first card (most relevant)
                return data[0]
            except requests.exceptions.RequestException as e:
                print(f"Request error: {e}")
                time.sleep(1)
        return None

    def matches_target_card(self, target_name: str, target_number: Optional[str],
                            result_name: str, result_info: Dict) -> Tuple[bool, int]:
        """
        Check if a PriceTracker result matches the target card.
        Returns (is_match, match_quality)
        """
        def base_name(name: str) -> str:
            return re.sub(r'\s*\([^)]+\)', '', name).strip().lower()

        match_quality = 0
        if base_name(target_name) != base_name(result_name):
            return False, 0

        match_quality += 10

        if target_number:
            result_number = str(result_info.get("number", ""))
            if target_number in result_number:
                match_quality += 20
            elif target_number.startswith("SV") and "SV" in result_number:
                match_quality += 15

        if target_name.lower() == result_name.lower():
            match_quality += 5

        return True, match_quality

    def search_price_tracker(self, target_name: str, target_number: Optional[str] = None) -> Optional[Dict]:
        """
        Search PriceTracker for best matching card.
        """
        headers = {"Authorization": f"Bearer {self.pricetracker_api_key}"}
        search_terms = [target_name]
        if target_number:
            search_terms.insert(0, f"{target_name} {target_number}")

        for term in search_terms:
            try:
                response = requests.get(
                    "https://www.pokemonpricetracker.com/api/v2/cards",
                    headers=headers,
                    params={"search": term, "limit": 10, "includeHistory": "false"},
                    timeout=30
                )
                self.api_calls_made += 1
                if response.status_code == 429:
                    print("Rate limit exceeded")
                    return None
                response.raise_for_status()
                results = response.json().get("data", [])
                if not results:
                    continue

                # Find best match
                best_match = None
                best_quality = 0
                for result in results:
                    result_name = result.get("name", "")
                    is_match, quality = self.matches_target_card(target_name, target_number, result_name, result)
                    if is_match and quality > best_quality:
                        best_match = result
                        best_quality = quality
                if best_match:
                    price = best_match.get("prices", {}).get("market")
                    return {"card": best_match, "price": price, "match_quality": best_quality}
            except requests.exceptions.RequestException as e:
                print(f"PriceTracker request error: {e}")
        return None

    def get_card_price(self, card_input: str) -> Optional[Dict]:
        """
        Complete lookup process using pokemontcg.io and PriceTracker.
        """
        name, number = self.parse_card_input(card_input)
        card_data = self.get_card_from_pokemontcg(name, number)
        if not card_data:
            print("Card not found on pokemontcg.io")
            return None

        exact_name = card_data.get("name")
        exact_number = card_data.get("number")
        return self.search_price_tracker(exact_name, exact_number)


# Simple interface
def get_card_price(card_input: str) -> Optional[float]:
    pricer = FinalCardPricer()
    result = pricer.get_card_price(card_input)
    if result and result.get("price") and result["price"] is not None:
        return float(result["price"])
    return None


# Example usage
if __name__ == "__main__":
    pricer = FinalCardPricer()
    test_cards = ["Rayquaza V 194/203", "Cynthia SV82"]
    for card in test_cards:
        price_data = pricer.get_card_price(card)
        if price_data and price_data.get("price"):
            print(f"{card}: ${price_data['price']}")
        else:
            print(f"{card}: Not found")
```
