```python

import os
import requests

#  Load API key from environment variable
API_KEY = os.getenv("POKE_PRICE_TRACKER_API_KEY")

# Check if API key is loaded
if not API_KEY:
    print("Error: API key not found. Please set POKE_PRICE_TRACKER_API_KEY.")
```

```python
#************************************************************
# OCR Output (example)
# Description: These are the OCR results from the card
#************************************************************
ocr_name = "Rayquaza C"
ocr_number = "8"   # From "8/147" we just need the 8

#************************************************************
# Method: search_card
# Input: name - card name, number - card number
# Description: Queries PokemonTCG.io API for a card matching
#              the given name and number. Returns first match info.
#************************************************************
def search_card(name, number):
    # Proper parameter format for the API
    params = {
        'q': f'name:"{name}" number:{number}'
    }
    
    url = "https://api.pokemontcg.io/v2/cards"
    response = requests.get(url, params=params)
    
    if response.status_code == 200: #Check if APi response suceeds
        data = response.json()
        if data["data"]:           # if any cards are found pick first
            # Pick the first match 
            card = data["data"][0]
            return {               # Return card info
                "id": card["id"],
                "name": card["name"],
                "number": card["number"],
                "set_name": card["set"]["name"],
                "set_id": card["set"]["id"],
                "series": card["set"]["series"],
                "rarity": card.get("rarity", "Unknown")
            }
        else:
            print("No card found.") # If card isnt found return 
            return None
    else:
        print("Error:", response.status_code, response.text) # API request failed
        print("Request URL:", response.url)  # Debug Info
        return None

#************************************************************
# Method: search_card_flexible
# Input: name - card name, number - card number
# Description: Attempts multiple variations to find
#              the card if standard search fails.
#************************************************************
def search_card_flexible(name, number):
    # Try multiple query variations
    queries = [
        f'name:"{name}" number:{number}',
        f'name:{name.replace(" ", "*")} number:{number}',
        f'name:{name} number:{number}',
    ]
    
    for query in queries:
        params = {'q': query}
        url = "https://api.pokemontcg.io/v2/cards"
        response = requests.get(url, params=params)
        
        if response.status_code == 200:   #Checking for API response
            data = response.json()
            if data["data"]:
                card = data["data"][0]
                return {
                    "id": card["id"],     # Return card info
                    "name": card["name"],
                    "number": card["number"],
                    "set_name": card["set"]["name"],
                    "set_id": card["set"]["id"],
                    "series": card["set"]["series"],
                    "rarity": card.get("rarity", "Unknown")
                }
        else:
            print(f"Query '{query}' failed:", response.status_code)
    
    print("No card found with any query variation.")
    return None

#************************************************************
# Test both approaches
# Description: Try standard search first, if fails go to flexible
#************************************************************
print("=== Standard Search ===")
card_info = search_card(ocr_name, ocr_number)

if not card_info:
    print("\n=== Flexible Search ===")
    card_info = search_card_flexible(ocr_name, ocr_number)

#************************************************************
# Display 
# Description: Prints card information 
#************************************************************
if card_info:
    print("\n=== Card Information ===")
    print("Card ID:", card_info["id"])
    print("Card Name:", card_info["name"])
    print("Card Number:", card_info["number"])
    print("Set Name:", card_info["set_name"])
    print("Series:", card_info["series"])
    print("Rarity:", card_info["rarity"])
else:
    print("\nNo card information found.")
```

```python
#************************************************************
# Method: get_card_price
# Input: card_name - string, card_number - string (optional), API_KEY - string
# Description: Queries PokemonPriceTracker API v2 using a targeted
#              search by card name (cheapest and most reliable).
#              Returns market price if found. Optionally checks
#              for card number to match specific card.
#************************************************************
def get_card_price(card_name, card_number=None, API_KEY=None):
    # Base URL for PokemonPriceTracker API
    base_url = "https://www.pokemonpricetracker.com/api/v2/cards"
    
    # Authorization header with Bearer token
    headers = {"Authorization": f"Bearer {API_KEY}"}

    # Parameters for targeted search
    params = {
        "search": card_name,      # Search specifically for the card name
        "limit": 1,               # Only fetch top 5 results to save API credits
        "includeHistory": "false" # We only need current price
    }

    try:
        #************************************************************
        #  Make API request
        #************************************************************
        response = requests.get(base_url, headers=headers, params=params)

        #************************************************************
        # Check if API response is successful
        #************************************************************
        if response.status_code == 200:
            data = response.json()          # Convert JSON response to Python dict
            cards = data.get("data", [])    # List of cards returned

            if not cards:
                print(f"No cards found for '{card_name}'.")
                return None

            #************************************************************
            # S Look for best match among returned cards
            #************************************************************
            best_match = None
            for card in cards:
                card_name_display = card.get("name", "")   # Name from API
                name_match = card_name.lower() in card_name_display.lower()  # Partial match

                #********************************************************
                # Check card number if provided
                #********************************************************
                number_match = False
                if card_number:
                    if f"{card_number}/" in card_name_display or f"- {card_number}" in card_name_display:
                        number_match = True

                #********************************************************
                # Decide if this card is the best match
                #********************************************************
                if name_match:
                    if number_match:      # Exact number match takes priority
                        best_match = card
                        break
                    elif not best_match:  # Keep first partial match as backup
                        best_match = card

            #************************************************************
            # Return price information if match found
            #************************************************************
            if best_match:
                return {
                    "name": best_match["name"],
                    "set": best_match.get("set", "Unknown"),
                    "price": best_match.get("prices", {}).get("market"),
                    "currency": "USD",
                    "rarity": best_match.get("rarity", "Unknown")
                }

            print(f"No matching card found for '{card_name}' in search results.")
            return None

        else:
            #************************************************************
            #: Handle API errors
            #************************************************************
            print(f"Error fetching price: {response.status_code}")
            print("Response:", response.text)
            return None

    except Exception as e: # Handle exceptions
        print(f"Exception occurred: {e}")
        return None


#************************************************************
# Example usage
#************************************************************
ocr_name = "Rayquaza C"
ocr_number = "8"

# Fetch price info from PokemonPriceTracker
price_info = get_card_price(ocr_name, card_number=ocr_number, API_KEY=API_KEY)

# Display results
if price_info:
    print(f"Card: {price_info['name']} ({price_info['set']})")
    print(f"Price: {price_info['price']} {price_info['currency']}")
    print(f"Rarity: {price_info['rarity']}")
else:
    print("Price information not available.")
```
